```python
# @title Default title text
!pip install -q scikit-learn pandas
import pandas as pd

base = "https://huggingface.co/datasets/ai4bharat/IndicSentiment/resolve/main/data"
frames = []
for split in ["validation", "test"]:
    url = f"{base}/{split}/hi.json"
    try:
        frames.append(pd.read_json(url, lines=True))
    except ValueError:
        frames.append(pd.read_json(url))

df = pd.concat(frames, ignore_index=True)
print(df.shape)
print(df.columns.tolist())
print(df.head(3))
```

```python
TEXT_COL = "INDIC REVIEW"
LABEL_COL = "LABEL"

df = df[[TEXT_COL, LABEL_COL]].dropna()
print(df.shape)
print(df[LABEL_COL].value_counts())
print(df[TEXT_COL].str.len().describe())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix

X_train, X_test, y_train, y_test = train_test_split(
    df[TEXT_COL], df[LABEL_COL], test_size=0.2, stratify=df[LABEL_COL], random_state=42)

# Hindi-safe tokenizer: split on whitespace/punctuation, keep vowel signs attached
vec = TfidfVectorizer(token_pattern=r"[^\s।,.!?\"'()\-:;]+", ngram_range=(1, 2), min_df=2)
Xtr = vec.fit_transform(X_train)   # learn vocabulary from training data only
Xte = vec.transform(X_test)

clf = LogisticRegression(max_iter=1000)
clf.fit(Xtr, y_train)
pred = clf.predict(Xte)

print(classification_report(y_test, pred))
print(confusion_matrix(y_test, pred))
```

```python
mask = (pred != y_test.values)
for text, true, p in list(zip(X_test[mask], y_test[mask], pred[mask]))[:10]:
    print(f"TRUE={true}  PRED={p}\n{text}\n")
```
